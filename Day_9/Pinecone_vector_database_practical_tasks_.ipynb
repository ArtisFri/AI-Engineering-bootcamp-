When you perform "Upsert data", add a list of your own text strings.

When you perform "Search", use of your own text string.

```python
import os, time

pc = Pinecone(api_key=os.environ["PINECONE_API_KEY"])

schema = (
    SchemaBuilder()
      .add_string_field("chunk_text", full_text_search={"language": "en"}, embed={})
      .build()
)
if not pc.indexes.exists(name="quickstart"):
    pc.indexes.create(name="quickstart", schema=schema)

while not pc.indexes.describe(name="quickstart").status.ready:
    time.sleep(2)

index = pc.index(name="quickstart")


index.documents.upsert(
    namespace="__default__",
    documents=[
        {"_id": "1", "chunk_text": "Solar panels and wind turbines help countries reduce carbon emissions", "category": "energy"},
        {"_id": "2", "chunk_text": "A healthy breakfast with fruit and whole grains gives you energy for the day.", "category": "food"},
        {"_id": "3", "chunk_text": "The football team won the nationals", "category": "sport"},
        {"_id": "4", "chunk_text": "Learning a programming language takes patience and regular practice on small projects.", "category": "Learning"},
        {"_id": "5", "chunk_text": "The city plans to build bicycle lanes and expand public transport to improve air quality.", "category": "energy"},
    ],
)

time.sleep(10)  


meaning = index.documents.search(
    namespace="__default__",
    top_k=3,
    score_by=[{"type": "embed", "fields": ["chunk_text"], "query": "how do i get clean power"}],
    include_fields=["chunk_text"],
)


```

Print search results.

```python
# place your code here
for m in meaning.matches:
    print(m.id, f"{m.score:.3f}", m.chunk_text)
```
